```python
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
documents = [
    Document(page_content="Artificial Intelligence is the simulation of human intelligence by computers."),
    Document(page_content="Machine Learning is a branch of Artificial Intelligence that allows computers to learn from data."),
    Document(page_content="Deep Learning uses neural networks with multiple layers to learn complex patterns."),
    Document(page_content="Natural Language Processing allows computers to understand and process human language."),
    Document(page_content="RAG stands for Retrieval-Augmented Generation. It retrieves relevant information before generating an answer."),
    Document(page_content="RAG can reduce hallucinations by providing relevant external information to the language model.")
]

print("Documents loaded:", len(documents))
```

### Initialize HuggingFaceEmbeddings

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embeddings created!")
```

### Create FAISS Vector Store

```python
vector_db = FAISS.from_documents(
    documents,
    embeddings
)

print("Vector database created!")
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embeddings created!")
```

```python
vector_db = FAISS.from_documents(
    documents,
    embeddings
)

print("Vector database created!")
```

```python
retriever = vector_db.as_retriever(
    search_kwargs={"k": 2}
)
```

```python
question = "What is RAG?"

results = retriever.invoke(question)

for i, doc in enumerate(results):
    print("\nResult", i + 1)
    print(doc.page_content)
```

```python
def rag(question):

    results = retriever.invoke(question)

    print("\nQuestion:")
    print(question)

    print("\nRetrieved Information:")

    for doc in results:
        print("-", doc.page_content)
```

```python
rag("What is deep Learning?")
```
